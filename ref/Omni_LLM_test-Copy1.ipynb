````python
import logging
import torch
from pathlib import Path
from transformers import Qwen3OmniMoeThinkerForConditionalGeneration, Qwen3OmniMoeProcessor


from qwen_omni_utils import process_mm_info

logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

def initialize_model(model_id: str):
    """Loads the Qwen3-Omni Thinker model for text-only output."""
    logging.info(f"Loading Thinker model: {model_id}...")
    
    processor = Qwen3OmniMoeProcessor.from_pretrained(model_id)
    
    model = Qwen3OmniMoeThinkerForConditionalGeneration.from_pretrained(
        model_id, 
        device_map="auto", 
        torch_dtype="auto", 
        attn_implementation="flash_attention_2"
    )
    
    model.eval()
    return processor, model

def generate_caption(audio_path: Path, processor, model) -> str:
    """Passes audio through the Thinker to generate a text caption."""
    try:
        conversations = [
            {
                "role": "user",
                "content": [
                    {"type": "audio", "audio_url": str(audio_path)},
                    # {"type": "text", "text": "Please provide a detailed description of the audio."}
                    {"type": "text", "text": SYSTEM_PROMPT_2}
                ]
            }
        ]
        
        # 1. Apply the chat template to get the raw string (DO NOT tokenize yet)
        text_prompt = processor.apply_chat_template(
            conversations, 
            add_generation_prompt=True, 
            tokenize=False 
        )
        
        # 2. Extract the physical audio arrays using the official Qwen utility
        audios, images, videos = process_mm_info(conversations, use_audio_in_video=False)
        
        # 3. Call the processor directly, keeping it on the CPU initially
        inputs = processor(
            text=text_prompt, 
            audio=audios, 
            images=images,
            videos=videos,
            return_tensors="pt", 
            padding=True
        )

        # 4. Safely move to GPU AND cast only float tensors (audio features) to match the model's precision
        inputs = {
            k: v.to(device=model.device, dtype=model.dtype) if v.is_floating_point() else v.to(model.device)
            for k, v in inputs.items()
        }
        # Qwen often uses <|im_end|> for chat turns and <|endoftext|> for raw generation.
        # Grabbing both ensures the model stops exactly when the caption is complete.
        stop_tokens = [processor.tokenizer.eos_token_id]
        
        im_end_token = processor.tokenizer.convert_tokens_to_ids("<|im_end|>")
        if im_end_token is not None:
            stop_tokens.append(im_end_token)
            
        with torch.no_grad():
            generated_ids = model.generate(**inputs, max_new_tokens=2560, do_sample=True, temperature=0.1,top_p=0.95, 
                                           eos_token_id=stop_tokens, pad_token_id=processor.tokenizer.pad_token_id)
        
        # Trim the input prompt tokens from the generated output
        generated_ids_trimmed = [
            out_ids[len(in_ids):] for in_ids, out_ids in zip(inputs["input_ids"], generated_ids)
        ]
        
        response = processor.batch_decode(
            generated_ids_trimmed, 
            skip_special_tokens=True, 
            clean_up_tokenization_spaces=False
        )[0]
        
        return response.strip()

    except Exception as e:
        # Keeping exc_info=True to catch any future backend parsing errors
        logging.error(f"Failed to process {audio_path}:", exc_info=True)
        return "Error generating caption."

def process_audio_directory(input_dir: str, output_dir: str, model_id: str):
    in_path = Path(input_dir)
    out_path = Path(output_dir)
    
    if not in_path.exists():
        raise FileNotFoundError(f"Input directory does not exist: {in_path}")

    valid_extensions = {'.wav', '.mp3', '.flac', '.ogg', '.m4a'}
    all_audio_files = [f for f in in_path.rglob('*') if f.suffix.lower() in valid_extensions]
    logging.info(f"Found {len(all_audio_files)} total audio files in dataset.")

    # Filter out files that already have a corresponding .txt output
    unprocessed_files = []
    for audio_file in all_audio_files:
        relative_path = audio_file.relative_to(in_path)
        target_text_path = (out_path / relative_path).with_suffix('.txt')
        
        if not target_text_path.exists():
            unprocessed_files.append((audio_file, target_text_path))

    skipped_count = len(all_audio_files) - len(unprocessed_files)
    if skipped_count > 0:
        logging.info(f"Skipping {skipped_count} files that already have .txt captions.")

    if not unprocessed_files:
        logging.info("All audio files in the directory are already captioned. Nothing to do!")
        return

    logging.info(f"Processing remaining {len(unprocessed_files)} files...")
    processor, model = initialize_model(model_id)
    
    for idx, (audio_file, target_text_path) in enumerate(unprocessed_files, 1):
            logging.info(f"[{idx}/{len(unprocessed_files)}] Processing: {audio_file.name}")
            
            # Create output subdirectory structure if needed
            target_text_path.parent.mkdir(parents=True, exist_ok=True)
            
            # Run inference
            caption = generate_caption(audio_file, processor, model)
            
            # Write caption to file
            with open(target_text_path, 'w', encoding='utf-8') as f:
                f.write(caption)

if __name__ == "__main__":
    INPUT_DATA_DIR = "/home/voice/code/duynlt1/YT_data/Nhat_samples/data/audios/"
    OUTPUT_DATA_DIR = "/home/voice/code/duynlt1/YT_data/Nhat_samples/data/captioned_audio_dataset_instruct_system_prompt"
    MODEL_IDENTIFIER = "/home/voice/data/voice/voice-models/Qwen3-Omni-30B-A3B-Instruct"
    
    process_audio_directory(INPUT_DATA_DIR, OUTPUT_DATA_DIR, MODEL_IDENTIFIER)
````
